1.Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

2.Loading Dataset

```python
df = pd.read_csv('https://raw.githubusercontent.com/olahiru176-svg/QUiz/main/emulsion_stability_data.csv')

print("Dataset Info:")
df.info()

print("\nFirst 12 rows of the dataset:")
display(df.head(12))
```

3.Part A                            Generate XY scatter plots using stability (%) as the y variable and other numerical variables as x. Compute the pearson correlation coefficient for each variable with stability.
evaluate which variable has the most strongest linear relationship


```python
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
numerical_cols.remove('Stability (%)') # Corrected column name

correlation_results = {}

plt.figure(figsize=(15, 5))
for i, col in enumerate(numerical_cols):
    plt.subplot(1, len(numerical_cols), i + 1)
    sns.scatterplot(x=df[col], y=df['Stability (%)']) # Corrected column name
    plt.title(f'Stability vs {col}')
    plt.xlabel(col)
    plt.ylabel('Stability (%)')

    correlation = df[col].corr(df['Stability (%)']) # Corrected column name
    correlation_results[col] = correlation
    print(f"Pearson correlation between {col} and Stability: {correlation:.4f}")

plt.tight_layout()
plt.show()

# Evaluate which variable has the strongest linear relationship
if correlation_results:
    strongest_variable = max(correlation_results, key=lambda k: abs(correlation_results[k]))
    strongest_correlation = correlation_results[strongest_variable]
    print(f"\nThe variable with the strongest linear relationship with Stability is '{strongest_variable}' with a Pearson correlation of {strongest_correlation:.4f}.")
else:
    print("No numerical variables found for correlation analysis.")
```

4.Part B   :Fit a multiple linear regression model:
Stability = b0 + b1*Speed + b2*Surfactant + b3*Time
Report the regression equation with coefficients rounded to 3 decimal places.


```python
from sklearn.linear_model import LinearRegression

X = df[['Speed (rpm)', 'Surfactant (%)', 'Time (min)']]
y = df['Stability (%)']

model = LinearRegression()
model.fit(X, y)

b0 = model.intercept_
b1, b2, b3 = model.coef_

print(f"Regression Equation:\nStability = {b0:.3f} + {b1:.3f}*Speed + {b2:.3f}*Surfactant + {b3:.3f}*Time")
```

5.Part D :Calculate and report:

    R² (coefficient of determination)

    RMSE (Root Mean Square Error)

    Create a residuals vs. predicted values plot.

```python
from sklearn.metrics import r2_score, mean_squared_error

# Get predictions from the fitted model
y_pred = model.predict(X)

# Calculate R-squared
r_squared = r2_score(y, y_pred)
print(f"R² (Coefficient of Determination): {r_squared:.3f}")

# Calculate RMSE (Root Mean Square Error)
rmse = np.sqrt(mean_squared_error(y, y_pred))
print(f"RMSE (Root Mean Square Error): {rmse:.3f}")

# Calculate residuals
residuals = y - y_pred

# Create residuals vs. predicted values plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_pred, y=residuals)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel('Predicted Stability (%)')
plt.ylabel('Residuals')
plt.title('Residuals vs. Predicted Values Plot')
plt.grid(True)
plt.show()
```
